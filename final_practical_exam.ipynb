### Clinical Premise
In precision oncology, pharmaceutical chemists test new molecular targeted cancer drugs across pre-clinical cancer cell line panels. The primary clinical measurement of drug efficacy is the **Half-Maximal Inhibitory Concentration ($\text{IC}_{50}$)** in micromolar ($\mu\text{M}$)—the drug concentration required to reduce cancer cell proliferation by $50\%$:
* **Low $\text{IC}_{50}$ ($<1.0\text{ }\mu\text{M}$):** Cancer cells are sensitive; the drug kills tumor cells at a low, safe dose.
* **High $\text{IC}_{50}$ ($>50.0\text{ }\mu\text{M}$):** Cancer cells are resistant; an unachievable toxic dose is required.

**4 Intuitive Predictors:**
1. `target_kinase_expression`: Expression of the enzyme inhibited by the drug (higher target = drug binds better $\longrightarrow$ lowers $\text{IC}_{50}$).
2. `efflux_transporter_activity`: Activity of membrane ABC transporters (pumps compound out of cell $\longrightarrow$ raises $\text{IC}_{50}$, conferring resistance).
3. `dna_repair_capacity`: Quantitative activity score of the tumor's nucleotide excision repair pathway (higher repair $\longrightarrow$ cancer cells survive damage $\longrightarrow$ raises $\text{IC}_{50}$).
4. `cell_doubling_time`: Proliferation rate of the tumor cell line in hours.

**Target Variable (Continuous):**
* `log_ic50`: Continuous log-transformed drug sensitivity metric ($\mu\text{M}$) where lower values denote greater drug potency.
